HuggingFace Endpoint

```python
#how to access higgingface models with api
from langchain_huggingface import HuggingFaceEndpoint,ChatHuggingFace
from langchain_core.messages import HumanMessage
```

```python
import os
os.environ['HUGGINGFACEHUB_API_TOKEN']=sec_key
```

```python
repo_id="mistralai/Mistral-7B-Instruct-v0.2"
llm=HuggingFaceEndpoint(repo_id=repo_id,task="conversational",max_new_tokens=200,
    temperature=0.5)

chat = ChatHuggingFace(llm=llm)

response = chat.invoke([
    HumanMessage(content="What is machine learning?")
])

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
question="Who won the cricket worldcup in 2019?"
template = """Question: {question}

Answer: Let's think step by step."""
prompt = PromptTemplate(template=template, input_variables=["question"])
```

```python
chain=prompt | chat
print(chain.invoke({"question":question}))
```

```python
from langchain_huggingface import HuggingFacePipeline
from transformers import pipeline,AutoModelForCausalLM,AutoTokenizer
```

```python
model_id="gpt2"
model=AutoModelForCausalLM.from_pretrained(model_id)
tokenizer=AutoTokenizer.from_pretrained(model_id)
```

```python
#pipeleine
pipe=pipeline("text-generation",model=model,tokenizer=tokenizer,max_new_tokens=200)
llm=HuggingFacePipeline(pipeline=pipe)
```

```python
llm.invoke("what is machine learning")
```

```python
gpu_llm=HuggingFacePipeline.from_model_id(model_id="gpt2",task="text-generation",device=0,pipeline_kwargs={"max_new_tokens":200})
```

```python
from langchain_core.prompts import PromptTemplate
template="""
Question:{question}
Answer:Let,s think step by step.
"""
prompt=PromptTemplate.from_template(template)
```

```python
chain=prompt | gpu_llm
```

```python
chain.invoke({"question":"what is Artifical Intelligance"})
```
